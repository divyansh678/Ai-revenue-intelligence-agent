```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("SuperStoreOrders.csv")
```

```python
df.head()
df.shape
```

```python
df.columns = (
    df.columns
      .str.strip()
      .str.lower()
      .str.replace(" ", "_")
      .str.replace("-", "_")
)
```

```python
df.columns.tolist()
```

```python
df[["order_date", "ship_date"]].head()
```

```python
df["order_date"] = pd.to_datetime(
    df["order_date"],
    format="mixed",
    dayfirst=True
)

df["ship_date"] = pd.to_datetime(
    df["ship_date"],
    format="mixed",
    dayfirst=True
)
```

```python
df[["order_date", "ship_date"]].head()
```

```python
null_count = df.isnull().sum()

null_percent = (
    df.isnull().sum() / len(df) * 100
).round(2)

missing = pd.DataFrame({
    "missing_values": null_count,
    "missing_percent": null_percent
})

missing[missing["missing_values"] > 0]
```

```python
df[["sales", "quantity", "discount", "profit"]].describe()
```

```python
df['sales'] = df['sales'].str.replace(',', '').astype(float)
```

```python
print("sales <= 0:", (df["sales"] <= 0).sum())
print("Quantity <= 0:", (df["quantity"] <= 0).sum())
print("Negative Profit:", (df["profit"] < 0).sum())
```

```python
invalid_shipping = df[
    df["ship_date"] < df["order_date"]
]

invalid_shipping.shape
```

```python
df["year"] = df["order_date"].dt.year

df["quarter"] = df["order_date"].dt.quarter

df["month_number"] = df["order_date"].dt.month

df["month"] = df["order_date"].dt.month_name()

df["year_month"] = df["order_date"].dt.to_period("M").astype(str)
```

```python
df[
    ["order_date", "year", "quarter",
     "month_number", "month", "year_month"]
].head()
```

```python
df["shipping_days"] = (
    df["ship_date"] - df["order_date"]
).dt.days
```

```python
df["shipping_days"].describe()
```

```python
df[
    ["order_date", "ship_date", "shipping_days"]
].head()
```

Profit margin


```python
df["profit_margin"] = (
    df["profit"] / df["sales"] * 100
).round(2)
```

```python
df[
    ["sales", "profit", "profit_margin"]
].head()
```

```python
df[
    [
        "year",
        "quarter",
        "month_number",
        "shipping_days",
        "profit_margin"
    ]
].describe()
```

```python
df["year"].value_counts().sort_index()
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nMissing values:")
print(df.isnull().sum()[df.isnull().sum() > 0])

print("\nDate range:")
print(df["order_date"].min(), "to", df["order_date"].max())
```

```python
df.to_csv("Processed_SuperStoreOrders.csv")
```
